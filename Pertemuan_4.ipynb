Sampling Distribution

````python
sample_means = []
````

````python
for i in range (1000):

  sample = np.random.choice(
      nilai,
      size= 2,
      replace = False
  )

  mean = np.mean(sample)

  sample_means.append(mean)
````

````python
#LINE CHART
x = [1, 2, 3, 4, 5]
y = [20, 30, 40, 10, 15]

plt.plot(x, y)
plt.show()
````

````python
#BAR PLOT

kategori = ["A", "B",  "C", "D"]
nilai = [80, 85, 90, 95]

plt.bar(kategori, nilai)
plt.show()
````

````python
#HISTOGRAM

nilai = [60, 70, 80, 90, 100]

plt.hist(nilai)
plt.show()
````

````python
populasi = np.random.normal(
    loc = 70, #mean
    scale = 5,#standar deviasi
    size = 1000 #banyaknya data
)
````

````python
plt.hist(populasi, bins=30)
plt.title("distribusi populasi")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show
````

Scipy

````python
from scipy import stats
````

````python
distribusi_normal = stats.norm.rvs(
    loc = 70,
    scale = 5,
    size = 1000
)

plt.hist(distribusi_normal, bins=30)
plt.title("DISTRIBUSI NORMAL")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show
````

````python
distribusi_exponensial = stats.expon.rvs(
    scale = 5, #std
    size = 1000 #banyaknya data
)

plt.hist(distribusi_exponensial, bins=30)
plt.title("DISTRIBUSI EXPONENSIAL")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show
````

````python
distribusi_poisson = stats.poisson.rvs(
    mu = 5, #rata-rata
    size = 1000 #banyaknya data
)

plt.hist(distribusi_poisson, bins=30)
plt.title("DISTRIBUSI POISSON")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show
````
